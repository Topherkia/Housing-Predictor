Loads an interactive map of **Melbourne** with [Folium](https://python-visualization.github.io/folium/) and shows every property from the chosen
[Melbourne Housing Snapshot](https://www.kaggle.com/datasets/dansbecker/melbourne-housing-snapshot) dataset, placed by its `Lattitude` / `Longtitude` columns.

* Markers are clustered – zoom in to see individual properties, click a marker for details.
* Marker colour = sale-price band (see legend, bottom left).
* Use the layer control (top right) to switch base maps, toggle House / Unit / Townhouse, or enable the price heat-map.

Run the cells from top to bottom (**Runtime → Run all**).

## 2. Choose the dataset

`raw` = original CSV, `processed` = cleaned, `processed_rm` = cleaned with unusually high prices removed. You can also paste a path or URL to your own CSV with the same columns.

```python
DATASET = "processed"  #@param ["raw", "processed", "processed_rm"] {allow-input: true}
SAMPLE_SIZE = 0        #@param {type:"integer"}
SHOW_HEATMAP = True    #@param {type:"boolean"}

df = load_dataset(DATASET)
if SAMPLE_SIZE and SAMPLE_SIZE < len(df):
    df = df.sample(SAMPLE_SIZE, random_state=42)
    print(f"Using a random sample of {len(df):,} properties")

df[["Suburb", "Address", "Type", "Price", "Lattitude", "Longtitude"]].head()
```

## 3. Show the map

```python
housing_map = build_map(df, dataset_label=DATASET_DESCRIPTIONS.get(DATASET, DATASET),
                        heatmap=SHOW_HEATMAP)
housing_map
```

## 4. (Optional) Save / download the map as HTML

```python
OUTPUT_FILE = "melbourne_housing_map.html"
housing_map.save(OUTPUT_FILE)
print("Saved", OUTPUT_FILE)

try:
    from google.colab import files  # only available on Colab
    files.download(OUTPUT_FILE)
except ImportError:
    print("Not running on Colab – open the HTML file in your browser.")
```
